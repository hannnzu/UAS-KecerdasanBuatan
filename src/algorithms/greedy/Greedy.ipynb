```python
# =====================================================================
# STEP 1: REPRESENTASI GRAFIK NETWORK (ADJACENCY LIST DENGAN BOBOT)
# =====================================================================
# Struktur data dictionary ini memetakan setiap node asal ke tetangganya
# beserta jarak (bobot) spesifik berdasarkan data jurnal.
graph = {
    'A': {'B': 3, 'F': 4},
    'B': {'A': 3, 'C': 14, 'E': 13, 'F': 7, 'J': 11, 'K': 16},
    'C': {'B': 14, 'D': 5, 'E': 6},
    'D': {'C': 5, 'E': 5},
    'E': {'B': 13, 'C': 6, 'D': 5, 'K': 3},
    'F': {'A': 4, 'B': 7, 'G': 7, 'J': 8},
    'G': {'F': 7, 'H': 4, 'J': 5},
    'H': {'G': 4, 'I': 9},
    'I': {'H': 9, 'N': 12},
    'J': {'B': 11, 'F': 8, 'G': 5, 'L': 3},
    'K': {'B': 16, 'E': 3, 'M': 6, 'N': 5},
    'L': {'J': 3, 'M': 7},
    'M': {'K': 6, 'L': 7, 'N': 3},
    'N': {'I': 12, 'K': 5, 'M': 3}
}
```

```python
# =====================================================================
# STEP 2: IMPLEMENTASI FUNGSI ALGORITMA GREEDY ROUTING
# =====================================================================
def greedy_routing(network, start_node, target_node):
    """
    Fungsi untuk mencari rute menggunakan prinsip Algoritma Greedy:
    "Ambil tetangga dengan jarak terkecil langsung di depan mata yang belum dikunjungi"
    """
    current_node = start_node
    path = [current_node]
    total_distance = 0
    visited = set([current_node]) # Menghindari perulangan/looping pada node yang sama
    steps = 0

    print(f"Mengeksekusi Routing Greedy: {start_node} -> {target_node}\n" + "-"*50)

    while current_node != target_node:
        steps += 1
        neighbors = network[current_node]

        best_neighbor = None
        min_distance = float('inf')

        # Evaluasi semua pilihan tetangga yang tersedia di depan mata
        for neighbor, distance in neighbors.items():
            if neighbor not in visited:
                if distance < min_distance:
                    min_distance = distance
                    best_neighbor = neighbor

        # Penanganan kondisi jika algoritma terjebak di jalan buntu
        if best_neighbor is None:
            print("Routing Gagal: Algoritma terjebak di dead-end (local maxima).")
            return None, float('inf'), steps

        # Akumulasikan jarak dan perbarui posisi node saat ini
        total_distance += min_distance
        current_node = best_neighbor
        path.append(current_node)
        visited.add(current_node)

        print(f"Proses Komputasi {steps}: Dari Node {path[-2]} memilih Node {current_node} (Jarak: {min_distance}) | Total Akumulasi Jarak: {total_distance}")

    print("-"*50)
    return path, total_distance, steps
```

```python
# =====================================================================
# STEP 2: IMPLEMENTASI FUNGSI ALGORITMA GREEDY ROUTING
# =====================================================================

# Mendefinisikan fungsi greedy_routing yang menerima 3 parameter: struktur graf (network), titik asal (start_node), dan titik tujuan (target_node)
def greedy_routing(network, start_node, target_node):
    # Dokumentasi fungsi bawaan (docstring) yang menjelaskan cara kerja fungsi ini
    """
    Fungsi untuk mencari rute menggunakan prinsip Algoritma Greedy:
    "Ambil tetangga dengan jarak terkecil langsung di depan mata yang belum dikunjungi"
    """
    # Menginisialisasi posisi saat ini (current_node) dengan titik asal/awal
    current_node = start_node
    # Membuat daftar (list) rute perjalanan dan langsung memasukkan node awal ke dalamnya
    path = [current_node]
    # Mengatur total akumulasi jarak awal menjadi 0
    total_distance = 0
    # Membuat himpunan (set) untuk mencatat node yang sudah dilewati agar tidak terjadi looping (diisi awal dengan start_node)
    visited = set([current_node]) # Menghindari perulangan/looping pada node yang sama
    # Menginisialisasi variabel penghitung langkah komputasi menjadi 0
    steps = 0

    # Mencetak batas awal yang menginformasikan bahwa proses pencarian rute dimulai
    print(f"Mengeksekusi Routing Greedy: {start_node} -> {target_node}\n" + "-"*50)

    # Memulai perulangan (loop) yang akan terus berjalan selama posisi saat ini belum sama dengan titik tujuan
    while current_node != target_node:
        # Menambah jumlah langkah sebesar 1 setiap kali iterasi (loop) berjalan
        steps += 1
        # Mengambil daftar node tetangga yang terhubung langsung dengan current_node dari dictionary network
        neighbors = network[current_node]
        print()

        # Menyiapkan variabel untuk menampung kandidat tetangga terdekat, default-nya None
        best_neighbor = None
        # Mengatur jarak minimum awal ke nilai tak terhingga (infinity) untuk memudahkan proses pencarian nilai terkecil
        min_distance = float('inf')

        # Evaluasi semua pilihan tetangga yang tersedia di depan mata
        # Melakukan iterasi/perulangan untuk memeriksa setiap node tetangga beserta jaraknya masing-masing
        for neighbor, distance in neighbors.items():
            # Mengecek apakah tetangga yang sedang dievaluasi ini belum pernah dikunjungi sebelumnya
            if neighbor not in visited:
                # Membandingkan jarak tetangga saat ini, apakah lebih kecil dari min_distance yang sedang tercatat
                if distance < min_distance:
                    # Jika lebih kecil, maka perbarui nilai min_distance dengan jarak tetangga tersebut
                    min_distance = distance
                    # Jadikan tetangga ini sebagai kandidat tetangga terbaik (terdekat) saat ini
                    best_neighbor = neighbor

        # Penanganan kondisi jika algoritma terjebak di jalan buntu
        # Mengecek apakah setelah loop tetangga di atas, tidak ada satu pun tetangga valid yang ditemukan (buntu)
        if best_neighbor is None:
            # Mencetak pesan peringatan ke layar bahwa algoritma terjebak di dead-end
            print("Routing Gagal: Algoritma terjebak di dead-end (local maxima).")
            # Menghentikan fungsi dan mengembalikan jalur kosong (None), jarak tak terhingga, dan langkah terakhir
            return None, float('inf'), steps

        # Akumulasikan jarak dan perbarui posisi node saat ini
        # Menambahkan jarak dari tetangga terpilih ke variabel total_distance
        total_distance += min_distance
        # Memperbarui posisi current_node ke titik tetangga terbaik yang baru saja ditemukan
        current_node = best_neighbor
        # Menambahkan titik baru ini ke dalam riwayat perjalanan (path)
        path.append(current_node)
        # Menandai titik baru ini ke dalam daftar visited agar tidak dikunjungi lagi di langkah berikutnya
        visited.add(current_node)

        # Mencetak rincian informasi pemindahan titik dan total jarak di langkah komputasi saat ini
        print(f"Proses Komputasi {steps}: Dari Node {path[-2]} memilih Node {current_node} (Jarak: {min_distance}) | Total Akumulasi Jarak: {total_distance}")

    # Mencetak garis pemisah untuk menandakan proses loop selesai atau tujuan berhasil dicapai
    print("-"*50)
    # Mengembalikan hasil rute utuh yang ditemukan, total jarak yang ditempuh, dan total langkah komputasinya
    return path, total_distance, steps
```
